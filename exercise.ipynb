```python
# Exercise 2: Movie Ticket Pricing (Logical Operators)
#Goal: Ask the user for their age (integer) and whether it's Wednesday (is_wednesday = input("Is today Wednesday? (y/n): ")).
# Regular price is $12
# Children (under 12) or Seniors (65 and older) pay $8
# Wednesday Special: If it is Wednesday, everyone gets a $2 discount off their ticket price!
# Input: Age: 10, Wednesday: y
# Expected Output: Your ticket price is: $6
# Input: Age: 25, Wednesday: n
# Expected Output: Your ticket price is: $12
# Input: Age: 25, Wednesday: y
# Expected Output: Your ticket price is: $10

age = int(input("What is your age?"))
is_wednesday = input("IS today Wednesday? (y/n):")

if age <= 12 or age >= 65:
    price = 8
elif age > 12 and age < 65:
    price = 12
else: 
    print("Input an age")

if is_wednesday == "y":
    price = price - 2
elif is_wednesday =="n":
    price = price

print(f"Your ticket ${price}")
```

```python
# Exercise 5: Phone Number Sanitizer & Checker
# Goal: Ask the user for a phone number that might 
# contain hyphens or spaces (e.g. 0812-3456-7890 or 0812 3456 7890).
# Remove all dashes "-" and spaces " " using .replace()
# Check if the remaining string:
# Contains only digits (.isdigit())
# Has a length between 10 and 13 characters (inclusive)
# Print the cleaned number if valid, or an error if invalid.
# Input: 0812-3456-7890
# Expected Output: Cleaned number: 081234567890 (Valid)
# Input: 0812-abc-7890
# Expected Output: Invalid: contains non-digit characters
# Input: 123
# Expected Output: Invalid: incorrect length

phnum = (input("What is your phone number?"))
fix = phnum.replace("-", "").replace(" ", "")

if not fix.isdigit():
    print("Invalid: contains non-digit characters")
elif not (10 <= len(fix) <= 13):
    print("Invalid: incorrect lenght")
else:
    print(f"Cleaned number: {fix} (valid)")
```

### Exercise 21 Radiation Oncology — Safe Working Distance (Inverse-Square Law)
##### $I_2 = I_1 \times \left(\frac{d_1}{d_2}\right)^2$

```python
def assign(variable):
    value = float(input(f"What is the value of {variable}:"))
    return value

I_1 = assign("I_1")
d_1 = assign("d_1")
d_2 = assign("d_2")

I_2 = round(I_1 * pow(d_1/d_2, 2), 2)

if I_2 > 2:
    print("CAUTION: Exceeds safe limit")
else: 
    print("SAFE: Within permitted dose rate")
```

### Exercise 22 Nuclear Medicine — Radiotracer Decay (Tc-99m)
##### Technetium-99m ($T_{1/2} = 6.0\text{ hours}$)
##### Exponential radioactive decay: $A = A_0 \times (\frac{1}{2})^{(t / T_{1/2})}$

```python
def assign(variable):
    value = float(input(f"What is the value of {variable}:"))
    return value

A_0 = 740.0
t = assign("t")
T_half = 6.0

A = round(A_0 * pow(1/2, t/T_half),3)

if A < 350.0:
    print(f"Remaining: {A} MBq -> INSUFFICIENT: Re-dose required")
else: 
    print(f"Remaining: {A} MBq -> APPROVED: Scan can proceed")
```

### Exercise 23 Arterial Stenosis & Flow Resistance (Poiseuille's Law)
##### vascular resistance $R$ is inversely proportional to the 4th power of radius: 
##### $R_{\text{ratio}} = \left(\frac{r_{\text{normal}}}{r_{\text{stenosed}}}\right)^4 = (\frac{1}{r_{\text{remaining}}})^4$

Ask for or set the percentage of radius remaining (e.g., $0.80$ means the lumen radius is $80%$ of normal, i.e., $20%$ plaque reduction).
###### $R_{\text{ratio}} < 2.0$: "Mild"
###### $2.0 \le R_{\text{ratio}} \le 5.0$: "Moderate Stenosis"
###### $R_{\text{ratio}} > 5.0$: "Critical: High Risk of Ischemia"

```python
def assign(variable):
    value = float(input(f"What is the value of {variable}:"))
    return value

r_remaining = assign("r_remaining")
R_ratio = round(pow(1/r_remaining, 4),3)

if R_ratio < 2: 
    status = "Mild"
elif (2 <= R_ratio <= 5):
    status = "Moderate Stenosis"
elif R_ratio > 5:
    status = "Critical: High Risk of Ischemia"
else: 
    print("Error")

print(f"Resistance increased by {R_ratio}x -> {status} ")
```

### Exercise 24 Radiology — Lead Shielding & HVL (Half-Value Layer)
##### $I = I_0 \times (0.5)^{(x / \text{HVL})}$ 
##### $\text{Blocked \%} = \left(1 - \frac{I}{I_0}\right) \times 100$
###### Beam HVL in lead is $0.5\text{ mm}$.

```python
def assign(variable):
    value = float(input(f"What is the value of {variable}:"))
    return value

x = assign("x")
HVL = 0.5

Intensity = 0.5 ** (x / HVL)
blocked = (1 - Intensity) * 100

if blocked <= 95:
    status = "Shielding Non-Compliant (Too thin)"
else:
    status = "Shielding Compliant"

print(f"Blocked {blocked}% -> {status}")
```

### Exercise 25 Diagnostic Ultrasound — Acoustic Impedance & Echo Reflection
##### $R = \left(\frac{Z_2 - Z_1}{Z_2 + Z_1}\right)^2$ 
##### $\text{Reflected \%} = R \times 100$
###### Muscle: $Z_1 = 1.70\text{ MRayls}$
###### Fat: $Z_2 = 1.38\text{ MRayls}$
###### Bone: $Z_{\text{bone}} = 6.00\text{ MRayls}$

###### Calculate $\text{Reflected \%}$

```python
impedances = {
    "muscle": 1.70,
    "fat": 1.38,
    "bone": 6.00,
    "air": 0.0004
}

def get_impedance(label):
    while True:
        choice = (input(f"Select material for {label}:").strip().lower())
        if choice in impedances:
            return impedances[choice]
        print("Invalid material. Please choose from the list.")

z1 = get_impedance("Z1")
z2 = get_impedance("Z2")
    
r = pow((z2 - z1)/ (z2 + z1), 2)
refl = round(r * 100, 2)

if refl > 40:
    status = "Strong Acoustic Shadowing (Cannot image past this boundary)"
else:
    status = "Good acoustic transmission window"

print(f"Reflection: {refl}% -> {status}")
```
