## Exercise 1

```python
# Exercise 1: Email Quick-Check
# Goal: Ask user email adress and validate:
# Must contain @ and must end with .com
# Input 1: alex@gmail.com, expected output is = Valid email adress
# Input 2: alexgmail.com or alexemail.net, expected output is = Invalid email adress

email = input("Input your email: ")

status = "Valid email adress" if email.find("@") >= 0 and email.find(".com") >= 0 else "Invalid email adress"
print(status)
```

## Exercise 2

```python
# Exercise 2: Movie Ticket Pricing (Logical Operators)
#Goal: Ask the user for their age (integer) and whether it's Wednesday (is_wednesday = input("Is today Wednesday? (y/n): ")).
# Regular price is $12
# Children (under 12) or Seniors (65 and older) pay $8
# Wednesday Special: If it is Wednesday, everyone gets a $2 discount off their ticket price!
# Input: Age: 10, Wednesday: y
# Expected Output: Your ticket price is: $6
# Input: Age: 25, Wednesday: n
# Expected Output: Your ticket price is: $12
# Input: Age: 25, Wednesday: y
# Expected Output: Your ticket price is: $10

age = int(input("What is your age?"))
is_wednesday = input("IS today Wednesday? (y/n):")

if age <= 12 or age >= 65:
    price = 8
elif age > 12 and age < 65:
    price = 12
else: 
    print("Input an age")

if is_wednesday == "y":
    price = price - 2
elif is_wednesday =="n":
    price = price

print(f"Your ticket ${price}")
```

## Exercise 3

```python
# Exercise 3: Grade Pass/Fail Tag (Ternary Expression)
# Goal: Ask the user for their test score (0 - 100).
# Use a single-line conditional expression (... if ... else ...) to determine:
# If score is 70 or above $\rightarrow$ "PASS"
# If score is below 70 $\rightarrow$ "FAIL"
# Print: "Result: [PASS/FAIL] with [score] points"
# Input: 85
# Expected Output: Result: PASS with 85 points
# Input: 55
# Expected Output: Result: FAIL with 55 points

score = int(input("What is your score: "))
status = "PASS" if score >= 70 else "FAIL" 
print(f"{status} with {score} points")
```

## Exercise 4

```python
# Exercise 4: Vowel Frequency Counter (String Methods)
# Goal: Ask the user for a word or phrase, and count how many times 
# the letter a appears (case-insensitive: both 'A' and 'a' should count).
# Hint: Convert the input to lowercase first!
# Input: Alabama
# Expected Output: The letter 'a' appears 4 times in 'alabama'
# Input: Python
# Expected Output: The letter 'a' appears 0 times in 'python'
 

word = input("Give me a word or phrase: ")
check = word.lower()
find_a = check.count("a")

print(f"The letter 'a' appears {find_a} times in '{word}")
```

## Exercise 5

```python
# Exercise 5: Phone Number Sanitizer & Checker
# Goal: Ask the user for a phone number that might 
# contain hyphens or spaces (e.g. 0812-3456-7890 or 0812 3456 7890).
# Remove all dashes "-" and spaces " " using .replace()
# Check if the remaining string:
# Contains only digits (.isdigit())
# Has a length between 10 and 13 characters (inclusive)
# Print the cleaned number if valid, or an error if invalid.
# Input: 0812-3456-7890
# Expected Output: Cleaned number: 081234567890 (Valid)
# Input: 0812-abc-7890
# Expected Output: Invalid: contains non-digit characters
# Input: 123
# Expected Output: Invalid: incorrect length

phnum = (input("What is your phone number?"))
fix = phnum.replace("-", "").replace(" ", "")

if not fix.isdigit():
    print("Invalid: contains non-digit characters")
elif not (10 <= len(fix) <= 13):
    print("Invalid: incorrect lenght")
else:
    print(f"Cleaned number: {fix} (valid)")
```

### Exercise 12 Free Fall Distance & Speed

```python
import math

# Free Fall Distance
t = float(input("Input time:"))
d = 1/2 * 9.81 * (t**2)

# Speed
v = math.sqrt(2 * 9.81 * d)

print(f"Fallen: {round(d,1)} m, Impact Speed: {round(v,1)} m/s")
```

### Exercise 14 Density & Buoyancy Check (Float or Sink?)

```python
m = float(input("Input mass:"))
V = float(input("Input Volume:"))

# Density
rho = m / V

# Buoyancy check
if rho == 1:
    print(f"Density: {round(rho,1)} g/cm^3 -> In water")
elif rho < 1:
    print(f"Density: {round(rho,1)} g/cm^3 -> Floats")
elif rho >= 1: 
    print(f"Density: {round(rho,1)} g/cm^3 -> Sinks")
```

### Exercise 18 pH Scale Classifier

```python
pH = float(input("Enter pH value:"))

if pH < 0 or pH > 14:
    print("Invalid pH range")
elif pH < 7:
    print("Acidic")
elif pH == 7: 
    print("Neutral")
elif pH > 7: 
    print("Alkaline")
else:
    print("Error")
```

### Exercise 20 Mass Percent of Solute
##### $\text{Mass \%} = \left(\frac{\text{solute mass}}{\text{solute mass} + \text{solvent mass}}\right) \times 100$

```python
solute_mass = float(input("Input solute mass:"))
solvent_mass = float(input("Input solvent mass:"))

mass_percentage = ((solute_mass) / (solute_mass + solvent_mass)) * 100
print(f"Concentration: {round(mass_percentage,2)} %")
```
